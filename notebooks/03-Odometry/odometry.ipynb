# 🚙 💻 Odométrie basée sur les encodeurs

L'odométrie consiste à mesurer la trajectoire, ou l'évolution de la pose avec temps, du robot.

On peut résoudre ce problème d'odométrie en utilisant les mesures des encodeurs de roue et un modèle de cinématique pour estimer l'évolution de la pose par une procédure itérative.

<p style="text-align:center;"><img src="../../assets/images/odometry/odometry-1.png" width="500" alt="odometry-1"></p>   

$$ x_{k+1} = x_k + \Delta x_k $$
$$ y_{k+1} = y_k + \Delta y_k $$
$$ \theta_{k+1} = \theta_k + \Delta \theta_k $$

On suppose que les conditions initiales ($x_0$, $y_0$, $\theta_0$) sont connues. Les incréments peuvent être calculés par :

1. **Détermination de la rotation de chaque roue grâce aux mesures de l'encodeur de roue**

$$\Delta \phi_k = N_k \cdot \alpha$$

où $N_k$ représente le nombre ticks mesuré par les encodeurs au cours du $k$-ième intervalle de temps, $\alpha = \frac{2 \pi}{N_{tot}}$ la rotation par tick, et $N_{tot}$ le nombre total de ticks par tour complet ($N_{tot} = 135$ pour les codeurs de roue utilisés). Cette relation est évaluée pour chaque roue, ce qui donne $\Delta \phi_{l,k}$ et $\Delta \phi_{r,k}$ pour les roues gauche et droite respectivement.

2. **Calcul de la distance totale parcourue par chaque roue**

<p style="text-align:center;"><img src="../../assets/images/odometry/odometry-d.png" width="250" alt="odometry-d"></p> 

En supposant que les rayons des roues soient identiques (égaux à $R$) pour les deux roues, la distance parcourue par chaque roue est donnée par :


$$ d_{l/r, k} = R \cdot \Delta \phi_{l/r,k}$$

3. **Déterminer la rotation et la distance parcourue par le robot (robot frame)**

<p style="text-align:center;"><img src="../../assets/images/odometry/odometry-2.png" width="250" alt="odometry-2"></p>    

En supposant qu'il n'y ait pas de glissement des roues du robot, nous pouvons déduire la distance parcourue par l'origine du frame du robot (point A) et la rotation du robot $\Delta \theta$:

$$ d_{A, k} = \frac{d_{r,k} + d_{l,k}}{2} $$
$$ \Delta \theta_{k} = \frac{d_{r,k} - d_{l,k}}{2L}$$

4. **Représentation du mouvement du robot dans le  frame world $O$**

<p style="text-align:center;"><img src="../../assets/images/odometry/odometry-3.png" width="250" alt="odometry-3"></p>

Enfin, nous pouvons exprimer le mouvement estimé dans le frame world et trouver :

$$ \Delta x_k = d_{A, k} \cos\theta_k $$
$$ \Delta y_k = d_{A, k} \sin\theta_k $$
